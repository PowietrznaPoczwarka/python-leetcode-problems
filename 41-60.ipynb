```python
# 41. first missing positive

# Given an unsorted integer array nums. Return the smallest positive integer that is not present in nums.
# You must implement an algorithm that runs in O(n) time and uses O(1) auxiliary space.

# Example 1:

# Input: nums = [1,2,0]
# Output: 3
# Explanation: The numbers in the range [1,2] are all in the array.
# Example 2:

# Input: nums = [3,4,-1,1]
# Output: 2
# Explanation: 1 is in the array but 2 is missing.
# Example 3:

# Input: nums = [7,8,9,11,12]
# Output: 1
# Explanation: The smallest positive integer 1 is missing.


# hint 3 kinda tells us we need two passes.
# the first pass can be a rearrange pass - each positive integer should be at its own index
# second one is a check pass, if a positive integer is not at its own index, tehn its missing

from typing import List

class Solution:
    def firstMissingPositive(self, nums: List[int]) -> int:
        n = len(nums)

        # rearrange pass
        for i in range(n):
            print(i, nums[i])
            print("here should be: ", i+1)
            # check if its a number to swap and whether swapping is even possible
            # and whether swapping candidates are different - nums[nums[i] - 1] != nums[i]
            # if a number is at correct position, they essentially point to the same int in the list, so we can move further
            while nums[i] > 0 and nums[i] < n+1 and nums[nums[i] - 1] != nums[i]:
                correct_idx = nums[i] - 1
                # print("swapping...")
                nums[i], nums[correct_idx] = nums[correct_idx], nums[i]
                
        print("nums after swapping: ", nums)

        # check pass
        for i in range(n):
            if nums[i] != i+1:
                print("first missing int: ", i+1)
                return i+1
            
        return nums[i] + 1

Solution().firstMissingPositive([1,2,3])
# Solution().firstMissingPositive([7,8,9,11,12])
```

```python
# 42
# Given n non-negative integers representing an elevation map where the width of each bar is 1, compute how much water it can trap after raining.

# Input: height = [0,1,0,2,1,0,1,3,2,1,2,1]
# Output: 6
# Explanation: The above elevation map (black section) is represented by array [0,1,0,2,1,0,1,3,2,1,2,1]. In this case, 6 units of rain water (blue section) are being trapped.

# intuition, go over the list, keep the rolling total of area, keep in memory the highest point,
# for each h after the highest point, check how much does it differ from the highest point, add as potential_area_to_add
# a water can be trapped, if the height was going down and now it goes up...

# if we reach new highest point, add potential area to area and reset potential area
# if we reach the end and there's some potential_area_to_add greater than 0, then repeat this function from the the index of highest point+1 and add the result (recursive function?)

# actually, instead of using a recursive function, we can simply run the same function but starting from the other side of the leftover list ( so that the highest point would be last, potential_area_to_add would be 0)

from typing import List

class Solution:
    def trap(self, height: List[int]) -> int:
        
        def calculate_water(array: List[int]) -> tuple[int, int]:
            total_water = 0
            potential_water = 0
            highest_point = 0
            highest_index = 0
            
            for i, h in enumerate(array):
                if h >= highest_point:
                    total_water += potential_water
                    potential_water = 0
                    
                    highest_point = h
                    highest_index = i
                    
                else:
                    potential_water += (highest_point - h)
                    
            return total_water, highest_index
            
        water_trapped, peak_idx = calculate_water(height)

        # check if there's anything else. If there is, run the same function but on the leftover list.
        leftover = height[peak_idx:]
        leftover.reverse()
        leftover_water, _ = calculate_water(leftover)
        
        return water_trapped + leftover_water

# print(Solution().trap([0,1,0,2,1,0,1,3,2,1,2,1]))
```
